#  Data Cleaning Ola Ride Booking Analytics

```python
#import libraries
import pandas as pd
import numpy as np
```

## Check Missing Values

```python
df.isnull().sum()
```

```python
# Fill null values
cols = [
    "Canceled_Rides_by_Customer",
    "Canceled_Rides_by_Driver",
    "Incomplete_Rides_Reason",
     "Payment_Method"
]

for col in cols:
    df[col] = df[col].fillna("Not Applicable")
```

```python
df[cols].isnull().sum()
```

```python
df["Incomplete_Rides"] = df["Incomplete_Rides"].fillna("No")
```

```python
df.isnull().sum()
```

```python
cols = [
    "V_TAT",
    "C_TAT",
    "Driver_Ratings",
    "Customer_Rating"
]

for col in cols:
    df[col] = df[col].fillna(0)
```

```python
df[cols].isnull().sum()
```

```python
df.isnull().sum()
```

##  Check duplicate Record

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
df.head()
```

## Cleaning Data Types

```python
df.dtypes
```

##  Booking Status Check

```python
df['Booking_Status'].value_counts()
```

##  Categorical Columns Check

```python
categorical_cols = [
    'Booking_Status',
    'Payment_Method',
    'Vehicle_Type',
    'Pickup_Location',
    'Drop_Location',
]

for col in categorical_cols:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
    
```

## Text Data Cleaning

```python
text_cols=df.select_dtypes(include='object').columns

for col in text_cols:
    df[col]=df[col].astype("string").str.strip()

print("data cleaning completed successfully")

```

```python
print(df.head())
```

```python
print(text_cols)
```

 ## Date and Time Cleaning

```python
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
```

```python
df["Time"]=pd.to_datetime(df['Time'].astype(str),errors="coerce").dt.time
```

```python
df["Date"] = pd.to_datetime(df["Date"]).dt.date
#df
```

## Check Numeric Columns

```python
numeric_cols=[
    'V_TAT',
    'C_TAT',
    'Booking_Value',
    'Ride_Distance',
    'Driver_Ratings',
    'Customer_Rating'
]
for col in numeric_cols:
    df[col]=pd.to_numeric(df[col],errors='coerce')


```

```python
df[numeric_cols].dtypes
```

## Add New Analytical Columns

```python
df["Month"] = pd.to_datetime(df["Date"]).dt.month_name()
```

```python
df["Day"] = pd.to_datetime(df["Date"]).dt.day_name()
```

```python
df["Hour"]=pd.to_datetime(df["Time"].astype(str),errors="coerce").dt.hour
```

```python
df.to_csv("Ola_Cleaned_Data.csv", index=False)

print("Cleaned dataset saved successfully")
```
