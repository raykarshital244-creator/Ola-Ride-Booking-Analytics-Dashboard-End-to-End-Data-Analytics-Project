#  Ola Ride Booking Analytics – Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("ola_cleaned_data.csv")
df.head()
```

```python
df.describe()
```

## Booking Status Analysis

```python
df["Booking_Status"].value_counts()
```

```python
df["Booking_Status"].value_counts(normalize = True)*100
```

```python
df["Booking_Status"].value_counts().plot(
kind="bar",
title="Booking Status Analysis"
)

plt.xlabel("Booking_Status")
plt.ylabel("number of bookings")
plt.show()
```

##  Peak Hour Analysis

```python
df["Hour"].value_counts().sort_index()
```

```python
df["Hour"].value_counts().sort_index().plot(
    kind="line",
    marker="o",
    title="Booking by Hour"
)

plt.xlabel("Hour")
plt.ylabel("Number of Bookings")
plt.xticks(range(0,24))
plt.grid()
plt.show()
```
